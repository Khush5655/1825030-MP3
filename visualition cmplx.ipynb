```python
import numpy as np
import matplotlib.pyplot as plt


def create_complex_grid(limit=2.0, points_per_axis=20):
    """Return a rectangular grid represented as complex numbers."""
    real_axis = np.linspace(-limit, limit, points_per_axis)
    imaginary_axis = np.linspace(-limit, limit, points_per_axis)

    real_grid, imaginary_grid = np.meshgrid(real_axis, imaginary_axis)
    return real_grid + 1j * imaginary_grid


def plot_mapping(axis, values, heading):
    """Plot the real and imaginary components of a complex array."""
    axis.scatter(
        values.real.ravel(),
        values.imag.ravel(),
        s=18,
        alpha=0.75
    )
    axis.set_title(heading, fontsize=11)
    axis.set_xlabel("Real component")
    axis.set_ylabel("Imaginary component")
    axis.axhline(0, linewidth=0.7)
    axis.axvline(0, linewidth=0.7)
    axis.grid(True, linestyle=":", alpha=0.65)
    axis.set_aspect("equal", adjustable="datalim")


def main():
    # Generate sample points and evaluate each transformation.
    complex_points = create_complex_grid()
    mappings = [
        ("Original complex plane", complex_points),
        ("Square mapping: w = z²", complex_points ** 2),
        ("Cube mapping: w = z³", complex_points ** 3),
        ("Exponential mapping: w = eᶻ", np.exp(complex_points)),
    ]

    figure, axes = plt.subplots(2, 2, figsize=(11, 8))
    figure.suptitle(
        "Geometric Visualisation of Complex Functions",
        fontsize=14,
        fontweight="semibold"
    )

    for panel, (heading, mapped_points) in zip(axes.flat, mappings):
        plot_mapping(panel, mapped_points, heading)

    figure.tight_layout(rect=(0, 0, 1, 0.95))
    plt.show()
```
