```python
import numpy as np
import matplotlib.pyplot as plt

# Coordinates of the point before rotation
point_x = 2
point_y = 1
rotation_degrees = 45
rotation_radians = np.deg2rad(rotation_degrees)

# Construct the two-dimensional rotation matrix
rotation_matrix = np.array([
    [np.cos(rotation_radians), -np.sin(rotation_radians)],
    [np.sin(rotation_radians),  np.cos(rotation_radians)]
])

original_vector = np.array([point_x, point_y])
rotated_vector = rotation_matrix @ original_vector
rotated_x, rotated_y = rotated_vector

# Compare the original and rotated vectors
plt.plot([0, point_x], [0, point_y], "o-", label="Original vector")
plt.plot([0, rotated_x], [0, rotated_y], "o-", label="Rotated vector")
plt.xlabel("X coordinate")
plt.ylabel("Y coordinate")
plt.title(f"Rotation by {rotation_degrees} degrees")
plt.legend()
plt.grid(True)
plt.axis("equal")
plt.show()

print("Coordinates after rotation:", rotated_vector)
```
