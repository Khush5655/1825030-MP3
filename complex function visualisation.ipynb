```python
import numpy as np
import matplotlib.pyplot as plt

# Generate points in the complex plane
real_axis = np.linspace(-2, 2, 20)
imaginary_axis = np.linspace(-2, 2, 20)
real_grid, imaginary_grid = np.meshgrid(real_axis, imaginary_axis)
complex_grid = real_grid + 1j * imaginary_grid

# Apply different complex mappings
square_mapping = complex_grid ** 2
cube_mapping = complex_grid ** 3
exponential_mapping = np.exp(complex_grid)

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Original complex plane
axes[0, 0].scatter(real_grid, imaginary_grid)
axes[0, 0].set_title("Original z-plane")

# Image under z squared
axes[0, 1].scatter(square_mapping.real, square_mapping.imag)
axes[0, 1].set_title("Mapping: w = z²")

# Image under z cubed
axes[1, 0].scatter(cube_mapping.real, cube_mapping.imag)
axes[1, 0].set_title("Mapping: w = z³")

# Image under the exponential function
axes[1, 1].scatter(exponential_mapping.real, exponential_mapping.imag)
axes[1, 1].set_title("Mapping: w = exp(z)")

for axis in axes.flat:
    axis.set_xlabel("Real part")
    axis.set_ylabel("Imaginary part")
    axis.grid(True)

fig.tight_layout()
plt.show()
```
