```python
import numpy as np
import matplotlib.pyplot as plt

# Values of the input sequence
sequence = [1, 2, 3, 4]
sequence_length = len(sequence)
transform_values = []

# Evaluate the DFT directly from its definition
for frequency_index in range(sequence_length):
    total = 0j
    for sample_index, sample_value in enumerate(sequence):
        phase = -2j * np.pi * frequency_index * sample_index / sequence_length
        total += sample_value * np.exp(phase)
    transform_values.append(total)

print("Discrete Fourier Transform:")
for index, value in enumerate(transform_values):
    print(f"X[{index}] = {value}")

magnitude = np.abs(transform_values)

figure, axes = plt.subplots(1, 2, figsize=(9, 4))
axes[0].stem(range(sequence_length), sequence)
axes[0].set_title("Input sequence")
axes[0].set_xlabel("Sample index")
axes[0].set_ylabel("Value")
axes[0].grid(True)

axes[1].stem(range(sequence_length), magnitude)
axes[1].set_title("Magnitude of DFT")
axes[1].set_xlabel("Frequency index")
axes[1].set_ylabel("Magnitude")
axes[1].grid(True)

figure.tight_layout()
plt.show()
```
