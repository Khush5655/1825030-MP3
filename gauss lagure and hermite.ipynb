```python
import numpy as np
import matplotlib.pyplot as plt

# Number of quadrature points
point_count = 10

# Gauss-Laguerre quadrature for the integral of x * exp(-x)
laguerre_nodes, laguerre_weights = np.polynomial.laguerre.laggauss(point_count)
laguerre_result = np.sum(laguerre_weights * laguerre_nodes)
laguerre_exact = 1.0

print("Gauss-Laguerre quadrature")
print("Numerical result:", laguerre_result)
print("Exact result:", laguerre_exact)

# Gauss-Hermite quadrature for the integral of exp(-x^2)
hermite_nodes, hermite_weights = np.polynomial.hermite.hermgauss(point_count)
hermite_result = np.sum(hermite_weights)
hermite_exact = np.sqrt(np.pi)

print("\nGauss-Hermite quadrature")
print("Numerical result:", hermite_result)
print("Exact result:", hermite_exact)

# Display the two weighted functions
laguerre_x = np.linspace(0, 8, 100)
laguerre_y = laguerre_x * np.exp(-laguerre_x)
hermite_x = np.linspace(-4, 4, 100)
hermite_y = np.exp(-(hermite_x ** 2))

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot(laguerre_x, laguerre_y)
axes[0].set_title("Gauss-Laguerre integrand")
axes[0].set_xlabel("x")
axes[0].set_ylabel("x exp(-x)")
axes[0].grid(True)

axes[1].plot(hermite_x, hermite_y)
axes[1].set_title("Gauss-Hermite integrand")
axes[1].set_xlabel("x")
axes[1].set_ylabel("exp(-x²)")
axes[1].grid(True)

fig.tight_layout()
plt.show()
```
