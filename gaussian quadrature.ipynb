```python
import numpy as np

# Integration interval
lower_limit = 0.0
upper_limit = 1.0

# Function to be integrated
def integrand(value):
    return value ** 2

# Two-point Gauss-Legendre nodes mapped to [lower_limit, upper_limit]
midpoint = (lower_limit + upper_limit) / 2
half_width = (upper_limit - lower_limit) / 2
node_offset = 1 / np.sqrt(3)

first_node = midpoint - half_width * node_offset
second_node = midpoint + half_width * node_offset

# Evaluate the quadrature formula
integral_estimate = half_width * (
    integrand(first_node) + integrand(second_node)
)

print("Approximate integral =", integral_estimate)
```
