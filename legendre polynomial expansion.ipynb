```python
import numpy as np
import matplotlib.pyplot as plt

def target_function(value):
    return value ** 2

# Choose the expansion degree and integration order
expansion_degree = 2
integration_order = 5

# Gauss-Legendre nodes and weights
quadrature_nodes, quadrature_weights = np.polynomial.legendre.leggauss(
    integration_order
)

# Find the Legendre-series coefficients
coefficients = []

for degree in range(expansion_degree + 1):
    polynomial_values = np.polynomial.legendre.legval(
        quadrature_nodes, [0] * degree + [1]
    )
    coefficient_integral = np.sum(
        quadrature_weights
        * target_function(quadrature_nodes)
        * polynomial_values
    )
    coefficient = (2 * degree + 1) * coefficient_integral / 2
    coefficients.append(coefficient)
    print(f"Coefficient a{degree} = {coefficient:.4f}")

# Construct the degree-two expansion
graph_x = np.linspace(-1, 1, 100)
approximation = (
    coefficients[0]
    + coefficients[1] * graph_x
    + coefficients[2] * (3 * graph_x ** 2 - 1) / 2
)
original_values = target_function(graph_x)

plt.plot(graph_x, original_values, label="Original function")
plt.plot(graph_x, approximation, "--", label="Legendre expansion")
plt.xlabel("x")
plt.ylabel("Function value")
plt.title("Legendre polynomial expansion")
plt.grid(True)
plt.legend()
plt.show()
```
