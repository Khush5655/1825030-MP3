```python

import numpy as np
import matplotlib.pyplot as plt

# Taking a complex number
real = 2
imag = 1

# 1. Translation
tx = 1
ty = 2
trans_real = real + tx
trans_imag = imag + ty

# 2. Enlargement (Scaling)
factor = 2
scale_real = real * factor
scale_imag = imag * factor

# 3. Rotation through 90 degrees
angle = np.radians(90)
rot_real = real * np.cos(angle) - imag * np.sin(angle)
rot_imag = real * np.sin(angle) + imag * np.cos(angle)

# 4. Reflection across the real axis
ref_real = real
ref_imag = -imag

# Printing the transformed coordinates
print("Original:", real, imag)
print("After Translation:", trans_real, trans_imag)
print("After Scaling:", scale_real, scale_imag)
print("After Rotation:", round(rot_real, 2), round(rot_imag, 2))
print("After Reflection:", ref_real, ref_imag)

# Drawing the points on the Argand plane
plt.figure(figsize=(8, 6))

plt.scatter(real, imag, color='red', label='Original')
plt.scatter(trans_real, trans_imag, color='blue', label='Translation')
plt.scatter(scale_real, scale_imag, color='green', label='Scaling')
plt.scatter(rot_real, rot_imag, color='purple', label='Rotation')
plt.scatter(ref_real, ref_imag, color='orange', label='Reflection')

plt.xlabel("Real Axis")
plt.ylabel("Imaginary Axis")
plt.title("Geometrical Transformations of a Complex Number")

plt.axhline(y=0)
plt.axvline(x=0)
plt.grid(True)
plt.legend()
plt.show()
```
