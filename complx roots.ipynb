```python
import numpy as np
import matplotlib.pyplot as plt

# Define the complex number and the order of the roots
real_part = 1
imaginary_part = 1
root_order = 3

# Convert the complex number into polar form
modulus = np.hypot(real_part, imaginary_part)
argument = np.arctan2(imaginary_part, real_part)
root_modulus = modulus ** (1 / root_order)

print("Roots in Cartesian and polar forms")

for root_index in range(root_order):
    root_angle = (argument + 2 * np.pi * root_index) / root_order
    root_real = root_modulus * np.cos(root_angle)
    root_imaginary = root_modulus * np.sin(root_angle)

    print(f"\nRoot {root_index + 1}")
    print(f"Cartesian form: {root_real:.4f} + ({root_imaginary:.4f})i")
    print(
        f"Polar form: {root_modulus:.4f} "
        f"[cos({root_angle:.4f}) + i sin({root_angle:.4f})]"
    )

    plt.scatter(root_real, root_imaginary, color="red")
    plt.annotate(str(root_index + 1), (root_real, root_imaginary))

plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)
plt.xlabel("Real part")
plt.ylabel("Imaginary part")
plt.title(f"{root_order}rd roots of a complex number")
plt.grid(True)
plt.axis("equal")
plt.show()
```
