```python
import numpy as np
import matplotlib.pyplot as plt

# Sample the function over a finite interval
sample_count = 100
position = np.linspace(-5, 5, sample_count)
signal = np.exp(-(position ** 2))

# Apply the Fast Fourier Transform
spectrum = np.fft.fft(signal)
frequency = np.fft.fftfreq(sample_count, d=position[1] - position[0])

# Centre the zero-frequency component
spectrum = np.fft.fftshift(spectrum)
frequency = np.fft.fftshift(frequency)
amplitude = np.abs(spectrum)

print("Fast Fourier Transform completed.")
print("First five transformed values:")
print(spectrum[:5])

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].plot(position, signal)
axes[0].set_title("Gaussian function")
axes[0].set_xlabel("Position")
axes[0].set_ylabel("Function value")
axes[0].grid(True)

axes[1].plot(frequency, amplitude)
axes[1].set_title("FFT magnitude spectrum")
axes[1].set_xlabel("Frequency")
axes[1].set_ylabel("Magnitude")
axes[1].grid(True)

fig.tight_layout()
plt.show()
```
